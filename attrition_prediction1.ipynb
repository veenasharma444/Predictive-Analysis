```python
# ==========================================================
# STEP 6: CREATE EMPLOYEE-LEVEL OT FEATURES
# ==========================================================

def create_ot_features(ot_df, prefix):
    
    df = ot_df.copy()

    # Normalize employee code
    df["_emp_code"] = normalize_emp_code(df["EMP CODE"])

    # Identify daily OT columns
    chargeable_cols = [
        col for col in df.columns
        if col.endswith("_CH")
    ]

    non_chargeable_cols = [
        col for col in df.columns
        if col.endswith("_NCH")
    ]

    total_cols = [
        col for col in df.columns
        if col.endswith("_TOT")
    ]

    # Convert OT columns to numeric
    for col in chargeable_cols + non_chargeable_cols + total_cols:
        df[col] = pd.to_numeric(df[col], errors="coerce").fillna(0)

    # Employee-level features
    features = pd.DataFrame({
        "_emp_code": df["_emp_code"],

        f"{prefix}_Chargeable_OT":
            df[chargeable_cols].sum(axis=1),

        f"{prefix}_NonChargeable_OT":
            df[non_chargeable_cols].sum(axis=1),

        f"{prefix}_Total_OT":
            df[total_cols].sum(axis=1),

        f"{prefix}_OT_Days":
            (df[total_cols] > 0).sum(axis=1),

        f"{prefix}_Avg_Daily_OT":
            df[total_cols].mean(axis=1),

        f"{prefix}_Max_Daily_OT":
            df[total_cols].max(axis=1)
    })

    # In case the source has multiple rows for the same employee
    features = (
        features
        .groupby("_emp_code", as_index=False)
        .agg({
            f"{prefix}_Chargeable_OT": "sum",
            f"{prefix}_NonChargeable_OT": "sum",
            f"{prefix}_Total_OT": "sum",
            f"{prefix}_OT_Days": "sum",
            f"{prefix}_Avg_Daily_OT": "mean",
            f"{prefix}_Max_Daily_OT": "max"
        })
    )

    return features


# ----------------------------------------------------------
# Create features from OT1
# ----------------------------------------------------------

ot1_features = create_ot_features(
    overtime_data1,
    "OT1"
)


# ----------------------------------------------------------
# Create features from OT4
# ----------------------------------------------------------

ot4_features = create_ot_features(
    overtime_data4,
    "OT4"
)



print("OT1 features shape:", ot1_features.shape)
print("OT4 features shape:", ot4_features.shape)

print("\nOT1 features:")
print(ot1_features.head())

print("\nOT4 features:")
print(ot4_features.head())
```

```python
# ==========================================================
# STEP 7: CREATE EMPLOYEE-LEVEL LEAVE FEATURES
# ==========================================================

leave = leave_data.copy()

# Normalize employee code
leave["_emp_code"] = normalize_emp_code(
    leave["Employee Code"]
)

# Convert important leave columns to numeric
leave_numeric_cols = [
    "Opening Balance (A)",
    "Accrued Till Date (B)",
    "Consumed Till Date (C)",
    "Availed Adjustment (D)",
    "Lapse (E)",
    "Balance Till Date (F) = A+B-C+D-E",
    "Future Dated Approved Leave (G)",
    "Balance Including Future Dated Approved Leave H (F-G)",
    "Leave Encashment(I)",
    "Carry Forward To Next Year J= (H-I)"
]

for col in leave_numeric_cols:
    leave[col] = pd.to_numeric(
        leave[col],
        errors="coerce"
    ).fillna(0)


# ----------------------------------------------------------
# Create employee-level leave features
# ----------------------------------------------------------

leave_features = (
    leave
    .groupby("_emp_code")
    .agg(
        Leave_Opening_Balance=(
            "Opening Balance (A)", "sum"
        ),

        Leave_Accrued=(
            "Accrued Till Date (B)", "sum"
        ),

        Leave_Consumed=(
            "Consumed Till Date (C)", "sum"
        ),

        Leave_Adjustment=(
            "Availed Adjustment (D)", "sum"
        ),

        Leave_Lapse=(
            "Lapse (E)", "sum"
        ),

        Leave_Balance=(
            "Balance Till Date (F) = A+B-C+D-E", "sum"
        ),

        Leave_Future_Approved=(
            "Future Dated Approved Leave (G)", "sum"
        ),

        Leave_Balance_After_Future=(
            "Balance Including Future Dated Approved Leave H (F-G)", "sum"
        ),

        Leave_Encashment=(
            "Leave Encashment(I)", "sum"
        ),

        Leave_Carry_Forward=(
            "Carry Forward To Next Year J= (H-I)", "sum"
        ),

        Leave_Record_Count=(
            "Leave Type", "count"
        ),

        Leave_Type_Count=(
            "Leave Type", "nunique"
        )
    )
    .reset_index()
)


print("Leave features shape:", leave_features.shape)

print("\nUnique employees in leave features:")
print(leave_features["_emp_code"].nunique())

print("\nLeave features:")
print(leave_features.head())
```

```python
# ==========================================================
# STEP 13A: CREATE OT2 AND OT3 FEATURES
# ==========================================================

def create_ot_summary_features(df, prefix):
    temp = df.copy()

    # Normalize employee code
    temp["_emp_code"] = normalize_emp_code(temp["EMP CODE"])

    # Convert OT columns to numeric
    for col in ["CHARGEABLE TOTAL", "NON CHARGEABLE TOTAL", "TOTAL"]:
        temp[col] = pd.to_numeric(temp[col], errors="coerce").fillna(0)

    # Aggregate at employee level
    features = (
        temp.groupby("_emp_code", dropna=True)
        .agg(
            **{
                f"{prefix}_Chargeable_OT": ("CHARGEABLE TOTAL", "sum"),
                f"{prefix}_NonChargeable_OT": ("NON CHARGEABLE TOTAL", "sum"),
                f"{prefix}_Total_OT": ("TOTAL", "sum"),
            }
        )
        .reset_index()
    )

    return features


# Create OT2 features
ot2_features = create_ot_summary_features(
    overtime_data2,
    "OT2"
)

# Create OT3 features
ot3_features = create_ot_summary_features(
    overtime_data3,
    "OT3"
)


print("OT2 features shape:", ot2_features.shape)
print("OT3 features shape:", ot3_features.shape)

print("\nOT2 unique employees:",
    ot2_features["_emp_code"].nunique())

print("OT3 unique employees:",
    ot3_features["_emp_code"].nunique())

print("\nOT2 features:")
print(ot2_features.head())

print("\nOT3 features:")
print(ot3_features.head())
```

```python

# ==========================================================
# STEP 24: PREPARE MODELING DATASET
# ==========================================================

import numpy as np
import pandas as pd

# ----------------------------------------------------------
# Start from clean features
# ----------------------------------------------------------

model_features_step24 = clean_features.copy()


# ----------------------------------------------------------
# Remove extremely sparse categorical fields
# ----------------------------------------------------------

sparse_cols = [
    "Point of Hire",
    "Point of Repatriation",
]

model_features_step24 = [
    c for c in model_features_step24
    if c not in sparse_cols
]


# ----------------------------------------------------------
# Create working dataframe
# ----------------------------------------------------------

model_data = model_base[
    model_features_step24 + ["Attrition"]
].copy()


# ==========================================================
# 1. CONVERT EFFECTIVE START DATE TO DATETIME
# ==========================================================

if "Effective Start Date" in model_data.columns:

    model_data["Effective Start Date"] = pd.to_datetime(
        model_data["Effective Start Date"],
        errors="coerce"
    )


# ==========================================================
# 2. CREATE AGE FROM BIRTH DATE
# ==========================================================

reference_date = pd.Timestamp("2026-08-07")

if "Birth Date" in model_data.columns:

    model_data["Age_Years"] = (
        reference_date - model_data["Birth Date"]
    ).dt.days / 365.25

    model_data["Age_Years"] = (
        model_data["Age_Years"]
        .clip(lower=18, upper=70)
    )


# ==========================================================
# 3. CREATE TENURE FROM JOIN DATE
# ==========================================================

if "Actual Join Date" in model_data.columns:

    model_data["Calculated_Tenure_Years"] = (
        reference_date - model_data["Actual Join Date"]
    ).dt.days / 365.25

    model_data["Calculated_Tenure_Years"] = (
        model_data["Calculated_Tenure_Years"]
        .clip(lower=0)
    )


# ==========================================================
# 4. CREATE CONFIRMATION STATUS
# ==========================================================

if "Confirmation Date" in model_data.columns:

    model_data["Is_Confirmed"] = (
        model_data["Confirmation Date"].notna()
    ).astype(int)


# ==========================================================
# 5. CREATE OT DATA AVAILABILITY FLAGS
# ==========================================================

ot_groups = [
    "OT1",
    "OT2",
    "OT3",
    "OT4",
]

for ot in ot_groups:

    total_col = f"{ot}_Total_OT"

    if total_col in model_data.columns:

        model_data[f"{ot}_Data_Available"] = (
            model_data[total_col].notna()
        ).astype(int)


# ==========================================================
# 6. CREATE FORM / SALARY DATA AVAILABILITY FLAG
# ==========================================================

if "Monthly Gross Salary" in model_data.columns:

    model_data["Salary_Data_Available"] = (
        model_data["Monthly Gross Salary"].notna()
    ).astype(int)


# ==========================================================
# 7. CREATE LEAVE DATA AVAILABILITY FLAG
# ==========================================================

if "Leave_Balance" in model_data.columns:

    model_data["Leave_Data_Available"] = (
        model_data["Leave_Balance"].notna()
    ).astype(int)


# ==========================================================
# 8. FILL OT MISSING VALUES WITH ZERO
# ==========================================================

ot_feature_cols = [
    c for c in model_data.columns
    if c.startswith("OT1_")
    or c.startswith("OT2_")
    or c.startswith("OT3_")
    or c.startswith("OT4_")
]

for c in ot_feature_cols:

    if "Data_Available" not in c:

        model_data[c] = model_data[c].fillna(0)


# ==========================================================
# 9. FILL LEAVE NUMERIC MISSING VALUES
# ==========================================================

leave_feature_cols = [
    c for c in model_data.columns
    if c.startswith("Leave_")
]

for c in leave_feature_cols:

    if c != "Leave_Data_Available":

        model_data[c] = model_data[c].fillna(0)


# ==========================================================
# 10. CHECK DATASET
# ==========================================================

print("=" * 70)
print("STEP 24: MODELING DATASET")
print("=" * 70)

print("Shape:", model_data.shape)

print("\nTarget distribution:")
print(model_data["Attrition"].value_counts())

print("\nTarget percentage:")
print(
    model_data["Attrition"].value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("\nMissing target:")
print(model_data["Attrition"].isna().sum())

print("\nOT availability:")

for ot in ot_groups:

    flag = f"{ot}_Data_Available"

    if flag in model_data.columns:

        print(
            f"{ot}:",
            model_data[flag].value_counts().to_dict()
        )

print("\nSalary availability:")

if "Salary_Data_Available" in model_data.columns:
    print(
        model_data["Salary_Data_Available"]
        .value_counts()
        .to_dict()
    )

print("\nLeave availability:")

if "Leave_Data_Available" in model_data.columns:
    print(
        model_data["Leave_Data_Available"]
        .value_counts()
        .to_dict()
    )
```

```python

# ==========================================================
# STEP 25: BUILD X AND y FOR RANDOM FOREST
# ==========================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 25: BUILD MODEL INPUTS")
print("=" * 70)


# ==========================================================
# 1. COPY MODEL DATA
# ==========================================================

rf_data = model_data.copy()


# ==========================================================
# 2. TARGET
# ==========================================================

y = rf_data["Attrition"].astype(int)

X = rf_data.drop(
    columns=["Attrition"]
).copy()


# ==========================================================
# 3. CONVERT DATE FEATURES INTO NUMERIC FEATURES
# ==========================================================

date_columns = [
    c for c in X.columns
    if pd.api.types.is_datetime64_any_dtype(X[c])
]


print("\nDate columns to transform:")

for c in date_columns:
    print(" -", c)


# Use a fixed reference date for reproducibility
reference_date = pd.Timestamp("2026-08-07")


for c in date_columns:

    # Convert date to number of days from reference date
    X[c + "_Days"] = (
        reference_date - X[c]
    ).dt.days

    # Missing date indicator
    X[c + "_Missing"] = (
        X[c].isna()
    ).astype(int)

    # Remove original datetime column
    X.drop(
        columns=[c],
        inplace=True
    )


# ==========================================================
# 4. CHECK EFFECTIVE START DATE
# ==========================================================

# It was originally object, but Step 24 converted it.
# If anything still remains as object but looks like a date,
# we will handle it later rather than silently encoding it.


# ==========================================================
# 5. IDENTIFY CATEGORICAL / NUMERIC FEATURES
# ==========================================================

categorical_features = [
    c for c in X.columns
    if (
        pd.api.types.is_object_dtype(X[c])
        or pd.api.types.is_string_dtype(X[c])
        or isinstance(X[c].dtype, pd.CategoricalDtype)
    )
]


numeric_features = [
    c for c in X.columns
    if c not in categorical_features
]


# ==========================================================
# 6. PRINT SUMMARY
# ==========================================================

print("\n" + "=" * 70)
print("MODEL INPUT SUMMARY")
print("=" * 70)

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nTarget distribution:")
print(y.value_counts())

print("\nTarget percentages:")
print(
    y.value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("\nCategorical features:", len(categorical_features))
print("Numeric features:", len(numeric_features))

print("\nCategorical columns:")

for i, c in enumerate(categorical_features, 1):
    print(f"{i:3}. {c}")

print("\nNumeric columns:")

for i, c in enumerate(numeric_features, 1):
    print(f"{i:3}. {c}")


# ==========================================================
# 7. CHECK REMAINING OBJECT COLUMNS
# ==========================================================

print("\n" + "=" * 70)
print("REMAINING OBJECT / STRING COLUMNS")
print("=" * 70)

for c in categorical_features:

    print(
        f"{c}: "
        f"dtype={X[c].dtype}, "
        f"unique={X[c].nunique(dropna=True)}, "
        f"missing={X[c].isna().sum()}"
    )


# ==========================================================
# 8. FINAL SAFETY CHECK
# ==========================================================

print("\n" + "=" * 70)
print("SAFETY CHECK")
print("=" * 70)

print("Target column present in X:", "Attrition" in X.columns)

print(
    "Employee identity columns accidentally present:",
    [
        c for c in [
            "Employee Code",
            "Employee_Code_Clean",
            "_emp_code"
        ]
        if c in X.columns
    ]
)

print(
    "Resignation leakage columns present:",
    [
        c for c in [
            "Resignation Date",
            "Last Working Date",
            "FNF Settlement Date"
        ]
        if c in X.columns
    ]
)
```

```python
# ==========================================================
# STEP 26: TRAIN / TEST SPLIT + PREPROCESSING
# ==========================================================

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

print("=" * 70)
print("STEP 26: TRAIN / TEST SPLIT + PREPROCESSING")
print("=" * 70)


# ==========================================================
# 1. TRAIN / TEST SPLIT
# ==========================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


# ==========================================================
# 2. VERIFY CLASS DISTRIBUTION
# ==========================================================

print("\nTRAINING SET")
print("Shape:", X_train.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTraining target percentage:")
print(
    y_train.value_counts(normalize=True)
    .mul(100)
    .round(2)
)


print("\n" + "-" * 70)

print("\nTEST SET")
print("Shape:", X_test.shape)

print("\nTest target distribution:")
print(y_test.value_counts())

print("\nTest target percentage:")
print(
    y_test.value_counts(normalize=True)
    .mul(100)
    .round(2)
)


# ==========================================================
# 3. PREPROCESSING
# ==========================================================

# Numeric pipeline
numeric_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        )
    ]
)


# Categorical pipeline
categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="constant",
                fill_value="Missing"
            )
        ),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                min_frequency=2
            )
        )
    ]
)


# ==========================================================
# 4. COMBINE PREPROCESSORS
# ==========================================================

preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            numeric_pipeline,
            numeric_features
        ),
        (
            "categorical",
            categorical_pipeline,
            categorical_features
        )
    ],
    remainder="drop"
)


# ==========================================================
# 5. FIT ONLY ON TRAINING DATA
# ==========================================================

X_train_processed = preprocessor.fit_transform(
    X_train
)

X_test_processed = preprocessor.transform(
    X_test
)


# ==========================================================
# 6. OUTPUT SHAPE
# ==========================================================

print("\n" + "=" * 70)
print("PREPROCESSING RESULT")
print("=" * 70)

print(
    "Original feature count:",
    X_train.shape[1]
)

print(
    "Processed training shape:",
    X_train_processed.shape
)

print(
    "Processed test shape:",
    X_test_processed.shape
)


# ==========================================================
# 7. SPARSITY CHECK
# ==========================================================

print("\nProcessed matrix type:")
print(type(X_train_processed))

if hasattr(X_train_processed, "nnz"):

    print(
        "Training non-zero values:",
        X_train_processed.nnz
    )

    print(
        "Training matrix density:",
        round(
            X_train_processed.nnz /
            (
                X_train_processed.shape[0]
                * X_train_processed.shape[1]
            ) * 100,
            4
        ),
        "%"
    )


# ==========================================================
# 8. FINAL CHECK
# ==========================================================

print("\n" + "=" * 70)
print("FINAL CHECK")
print("=" * 70)

print(
    "Train rows:",
    len(X_train)
)

print(
    "Test rows:",
    len(X_test)
)

print(
    "Train attrition cases:",
    int(y_train.sum())
)

print(
    "Test attrition cases:",
    int(y_test.sum())
)

print(
    "Preprocessor fitted successfully: YES"
)
```

```python
# ==========================================================
# STEP 26.1: FIX MIXED CATEGORICAL VALUES
# ==========================================================

import pandas as pd
import numpy as np

# Make copies
X_train_fixed = X_train.copy()
X_test_fixed = X_test.copy()

# ----------------------------------------------------------
# 1. Clean categorical columns
# ----------------------------------------------------------

for col in categorical_features:

    # Convert everything to string safely.
    # Missing values are explicitly replaced before conversion.
    X_train_fixed[col] = (
        X_train_fixed[col]
        .astype("object")
        .where(X_train_fixed[col].notna(), "MISSING")
        .astype(str)
    )

    X_test_fixed[col] = (
        X_test_fixed[col]
        .astype("object")
        .where(X_test_fixed[col].notna(), "MISSING")
        .astype(str)
    )


# ----------------------------------------------------------
# 2. Clean numeric columns
# ----------------------------------------------------------

for col in numeric_features:

    X_train_fixed[col] = pd.to_numeric(
        X_train_fixed[col],
        errors="coerce"
    )

    X_test_fixed[col] = pd.to_numeric(
        X_test_fixed[col],
        errors="coerce"
    )


print("Categorical values fixed successfully.")
print("Training shape:", X_train_fixed.shape)
print("Test shape:", X_test_fixed.shape)


# ==========================================================
# 3. PREPROCESSOR
# ==========================================================

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer


categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="constant",
                fill_value="MISSING"
            )
        ),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=True
            )
        )
    ]
)


numeric_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        )
    ]
)


preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            categorical_pipeline,
            categorical_features
        ),
        (
            "numeric",
            numeric_pipeline,
            numeric_features
        )
    ],
    remainder="drop"
)


# ==========================================================
# 4. FIT ONLY ON TRAINING DATA
# ==========================================================

X_train_processed = preprocessor.fit_transform(
    X_train_fixed
)

X_test_processed = preprocessor.transform(
    X_test_fixed
)


# ==========================================================
# 5. RESULTS
# ==========================================================

print("\n" + "=" * 60)
print("STEP 26.1: PREPROCESSING SUCCESS")
print("=" * 60)

print("Original X train shape:", X_train.shape)
print("Processed X train shape:", X_train_processed.shape)

print("Original X test shape:", X_test.shape)
print("Processed X test shape:", X_test_processed.shape)

print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
# ==========================================================
# STEP 27: RANDOM FOREST MODEL
# ==========================================================

from sklearn.ensemble import RandomForestClassifier

# Create Random Forest
rf_model = RandomForestClassifier(
    n_estimators=300,
    max_depth=None,
    min_samples_split=10,
    min_samples_leaf=3,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

print("Training Random Forest...")

# Train
rf_model.fit(
    X_train_processed,
    y_train
)

print("\n" + "=" * 60)
print("STEP 27: RANDOM FOREST TRAINING SUCCESS")
print("=" * 60)

print("Number of trees:", rf_model.n_estimators)
print("Training samples:", X_train_processed.shape[0])
print("Processed features:", X_train_processed.shape[1])
```

```python
# ==========================================================
# STEP 29: RANDOM FOREST FEATURE IMPORTANCE
# ==========================================================

import pandas as pd
import numpy as np

print("=" * 60)
print("STEP 29: RANDOM FOREST FEATURE IMPORTANCE")
print("=" * 60)


# ==========================================================
# 1. GET FEATURE IMPORTANCE FROM RANDOM FOREST
# ==========================================================

encoded_importance = rf_model.feature_importances_

print("\nNumber of encoded features:",
      len(encoded_importance))


# ==========================================================
# 2. GET ENCODED FEATURE NAMES
# ==========================================================

encoded_feature_names = preprocessor.get_feature_names_out()

print("Number of encoded feature names:",
      len(encoded_feature_names))


# Safety check
assert len(encoded_importance) == len(encoded_feature_names), \
    "Feature importance and feature names count do not match."


# ==========================================================
# 3. CREATE ENCODED IMPORTANCE DATAFRAME
# ==========================================================

importance_df = pd.DataFrame({
    "Encoded_Feature": encoded_feature_names,
    "Importance": encoded_importance
})


# ==========================================================
# 4. MAP ENCODED FEATURES BACK TO ORIGINAL FEATURES
# ==========================================================

def get_original_feature(encoded_name):

    # Remove ColumnTransformer prefix
    if "__" in encoded_name:
        encoded_name = encoded_name.split("__", 1)[1]

    # Check categorical features first
    for col in categorical_features:
        if encoded_name == col or encoded_name.startswith(col + "_"):
            return col

    # Check numeric features
    for col in numeric_features:
        if encoded_name == col:
            return col

    # Fallback
    return encoded_name


importance_df["Original_Feature"] = (
    importance_df["Encoded_Feature"]
    .apply(get_original_feature)
)


# ==========================================================
# 5. AGGREGATE IMPORTANCE
# ==========================================================

feature_importance = (
    importance_df
    .groupby("Original_Feature", as_index=False)["Importance"]
    .sum()
    .sort_values(
        "Importance",
        ascending=False
    )
    .reset_index(drop=True)
)


# ==========================================================
# 6. CONVERT TO PERCENTAGE
# ==========================================================

feature_importance["Importance_%"] = (
    feature_importance["Importance"] * 100
)


# ==========================================================
# 7. ADD RANK
# ==========================================================

feature_importance.insert(
    0,
    "Rank",
    range(1, len(feature_importance) + 1)
)


# ==========================================================
# 8. DISPLAY TOP 30 PREDICTIVE FACTORS
# ==========================================================

print("\n" + "=" * 60)
print("TOP 30 PREDICTIVE FACTORS")
print("=" * 60)

print(
    feature_importance[
        [
            "Rank",
            "Original_Feature",
            "Importance",
            "Importance_%"
        ]
    ].head(30).to_string(index=False)
)


# ==========================================================
# 9. CHECK TOTAL IMPORTANCE
# ==========================================================

print("\n" + "=" * 60)
print("IMPORTANCE CHECK")
print("=" * 60)

print(
    "Total encoded importance:",
    encoded_importance.sum()
)

print(
    "Total aggregated importance:",
    feature_importance["Importance"].sum()
)


# ==========================================================
# 10. SAVE FEATURE IMPORTANCE
# ==========================================================

feature_importance.to_excel(
    "random_forest_feature_importance.xlsx",
    index=False
)

print("\nSaved:")
print("random_forest_feature_importance.xlsx")
```

```python
# ============================================================
# STEP 35: FIX MALFORMED EFFECTIVE START DATE
# ============================================================

import pandas as pd
import numpy as np
from datetime import datetime

col = "Effective Start Date"

print("=" * 60)
print("FIXING EFFECTIVE START DATE")
print("=" * 60)

# ------------------------------------------------------------
# 1. Find suspicious year values
# ------------------------------------------------------------

suspicious = []

for idx, value in model_data_clean[col].items():
    if pd.isna(value):
        continue

    text = str(value)

    # Detect years such as 0025, 0024, etc.
    if any(f"-{year:04d}-" in text for year in range(1, 100)):
        suspicious.append((idx, value))

print(f"\nSuspicious date values found: {len(suspicious)}")

for idx, value in suspicious[:20]:
    print(f"Index: {idx} | Value: {value}")


# ------------------------------------------------------------
# 2. Correct the known malformed date
# ------------------------------------------------------------

for idx, value in suspicious:
    text = str(value)

    if text == "01-DEC-0025":
        model_data_clean.at[idx, col] = datetime(2025, 12, 1)

        print(
            f"\nCorrected index {idx}: "
            f"{value} -> 2025-12-01"
        )


# ------------------------------------------------------------
# 3. Convert the complete column safely
# ------------------------------------------------------------

model_data_clean[col] = pd.to_datetime(
    model_data_clean[col],
    errors="coerce"
)


# ------------------------------------------------------------
# 4. Validation
# ------------------------------------------------------------

print("\nFinal dtype:")
print(model_data_clean[col].dtype)

print("\nMissing dates after conversion:")
print(model_data_clean[col].isna().sum())

print("\nMinimum date:")
print(model_data_clean[col].min())

print("\nMaximum date:")
print(model_data_clean[col].max())

print("\nDate conversion completed successfully.")
```

```python
# ============================================================
# STEP 35B: FIX YEAR 0025 IN EFFECTIVE START DATE
# ============================================================

col = "Effective Start Date"

print("=" * 60)
print("CHECKING FOR INVALID YEAR VALUES")
print("=" * 60)

# Find dates where year is less than 1900
invalid_mask = model_data_clean[col].notna() & (
    model_data_clean[col].dt.year < 1900
)

print(f"\nInvalid date rows found: {invalid_mask.sum()}")

if invalid_mask.sum() > 0:
    print("\nInvalid dates found:")
    print(model_data_clean.loc[invalid_mask, col].value_counts())

    # Specifically fix year 0025 -> 2025
    def fix_invalid_year(x):
        if pd.isna(x):
            return x

        if x.year == 25:
            return x.replace(year=2025)

        return x

    model_data_clean[col] = model_data_clean[col].apply(fix_invalid_year)

    print("\nCorrected year 0025 -> 2025")

# Convert to standard pandas datetime
model_data_clean[col] = pd.to_datetime(
    model_data_clean[col],
    errors="coerce"
)

print("\n" + "=" * 60)
print("VALIDATION")
print("=" * 60)

print("\nDtype:")
print(model_data_clean[col].dtype)

print("\nMinimum date:")
print(model_data_clean[col].min())

print("\nMaximum date:")
print(model_data_clean[col].max())

print("\nDates before 1900:")
print(
    (model_data_clean[col].notna() &
     (model_data_clean[col].dt.year < 1900)).sum()
)

print("\nMissing dates:")
print(model_data_clean[col].isna().sum())

print("\nDate correction completed successfully.")
```

```python
# ============================================================
# STEP 39: FEATURE CARDINALITY & IDENTIFIER AUDIT
# ============================================================

print("=" * 60)
print("STEP 39: FEATURE CARDINALITY & IDENTIFIER AUDIT")
print("=" * 60)


# ------------------------------------------------------------
# 1. Unique values in categorical columns
# ------------------------------------------------------------

categorical_cols = X_train.select_dtypes(
    exclude=["number"]
).columns.tolist()

categorical_cardinality = pd.DataFrame({
    "Column": categorical_cols,
    "Unique_Count": [
        X_train[col].nunique(dropna=True)
        for col in categorical_cols
    ],
    "Missing_Count": [
        X_train[col].isna().sum()
        for col in categorical_cols
    ]
})

categorical_cardinality["Unique_%"] = (
    categorical_cardinality["Unique_Count"]
    / len(X_train) * 100
)

categorical_cardinality = categorical_cardinality.sort_values(
    "Unique_Count",
    ascending=False
)


print("\n" + "=" * 60)
print("CATEGORICAL COLUMN CARDINALITY")
print("=" * 60)

print(categorical_cardinality.to_string(index=False))


# ------------------------------------------------------------
# 2. Numeric columns with very high cardinality
# ------------------------------------------------------------

numeric_cols = X_train.select_dtypes(
    include=["number"]
).columns.tolist()

numeric_cardinality = pd.DataFrame({
    "Column": numeric_cols,
    "Unique_Count": [
        X_train[col].nunique(dropna=True)
        for col in numeric_cols
    ],
    "Missing_Count": [
        X_train[col].isna().sum()
        for col in numeric_cols
    ]
})

numeric_cardinality = numeric_cardinality.sort_values(
    "Unique_Count",
    ascending=False
)


print("\n" + "=" * 60)
print("NUMERIC COLUMN CARDINALITY")
print("=" * 60)

print(numeric_cardinality.to_string(index=False))


# ------------------------------------------------------------
# 3. Possible identifier-like columns
# ------------------------------------------------------------

possible_id_cols = []

for col in X_train.columns:

    unique_count = X_train[col].nunique(dropna=True)
    unique_ratio = unique_count / len(X_train)

    col_lower = col.lower()

    if (
        unique_ratio > 0.80
        or
        any(word in col_lower for word in [
            "employee code",
            "employee id",
            "employee no",
            "emp code",
            "emp id",
            "person id",
            "user id"
        ])
    ):
        possible_id_cols.append(col)


print("\n" + "=" * 60)
print("POSSIBLE IDENTIFIER-LIKE COLUMNS")
print("=" * 60)

if possible_id_cols:
    for col in possible_id_cols:
        print(
            f"{col:<40} "
            f"unique={X_train[col].nunique(dropna=True)}"
        )
else:
    print("No obvious identifier-like columns found.")


# ------------------------------------------------------------
# 4. Very high-cardinality categorical columns
# ------------------------------------------------------------

high_cardinality_cat = categorical_cardinality[
    categorical_cardinality["Unique_%"] > 5
]

print("\n" + "=" * 60)
print("HIGH-CARDINALITY CATEGORICAL COLUMNS")
print("(More than 5% unique values)")
print("=" * 60)

if len(high_cardinality_cat) > 0:
    print(
        high_cardinality_cat.to_string(index=False)
    )
else:
    print("None found.")


print("\n" + "=" * 60)
print("STEP 39 COMPLETE")
print("=" * 60)
```

```python

# ============================================================
# STEP 40: REMOVE LEAKAGE & NON-INFORMATIVE FEATURES
# ============================================================

print("=" * 60)
print("STEP 40: FEATURE CLEANING BEFORE PREPROCESSING")
print("=" * 60)


# ------------------------------------------------------------
# 1. Features that can contain future information
# ------------------------------------------------------------

leakage_columns = [
    "Confirmation Date_Days",
    "Conf. Date (Period Master)_Days",
]


# ------------------------------------------------------------
# 2. Non-informative / constant feature
# ------------------------------------------------------------

constant_columns = [
    "Source of Hire",
]


# ------------------------------------------------------------
# 3. Combine columns to remove
# ------------------------------------------------------------

columns_to_drop = leakage_columns + constant_columns

# Keep only columns that actually exist
columns_to_drop = [
    col for col in columns_to_drop
    if col in X_train.columns
]

print("\nColumns to remove:")

for col in columns_to_drop:
    print("-", col)


# ------------------------------------------------------------
# 4. Remove from train and test
# ------------------------------------------------------------

X_train_clean = X_train.drop(
    columns=columns_to_drop
).copy()

X_test_clean = X_test.drop(
    columns=columns_to_drop
).copy()


# ------------------------------------------------------------
# 5. Verify
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("AFTER FEATURE REMOVAL")
print("=" * 60)

print("\nOriginal number of features:", X_train.shape[1])

print("Removed features:", len(columns_to_drop))

print("Remaining features:", X_train_clean.shape[1])

print("\nX_train_clean shape:",
      X_train_clean.shape)

print("X_test_clean shape:",
      X_test_clean.shape)


# ------------------------------------------------------------
# 6. Check for remaining datetime columns
# ------------------------------------------------------------

remaining_datetime = X_train_clean.select_dtypes(
    include=["datetime", "datetimetz"]
).columns.tolist()

print("\nRemaining datetime columns:")
print(remaining_datetime)


# ------------------------------------------------------------
# 7. Check target is not present
# ------------------------------------------------------------

print("\nTarget present in X_train_clean:",
      "Attrition" in X_train_clean.columns)


print("\n" + "=" * 60)
print("STEP 40 SUCCESS")
print("=" * 60)
```

```python
# ============================================================
# STEP 41: PREPROCESSING PIPELINE
# ============================================================

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

print("=" * 60)
print("STEP 41: PREPROCESSING")
print("=" * 60)


# ------------------------------------------------------------
# 1. Identify numeric and categorical columns
# ------------------------------------------------------------

numeric_features = X_train_clean.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_features = X_train_clean.select_dtypes(
    exclude=["number"]
).columns.tolist()


print("\nNumeric features:", len(numeric_features))
print("Categorical features:", len(categorical_features))


# ------------------------------------------------------------
# 2. FIX MIXED DATA TYPES IN CATEGORICAL COLUMNS
# ------------------------------------------------------------
# Some categorical columns contain both numbers and strings.
# Convert all non-missing categorical values to strings so that
# OneHotEncoder receives a consistent data type.

for col in categorical_features:

    X_train_clean[col] = X_train_clean[col].apply(
        lambda x: str(x) if pd.notna(x) else np.nan
    )

    X_test_clean[col] = X_test_clean[col].apply(
        lambda x: str(x) if pd.notna(x) else np.nan
    )


print("\nCategorical data types standardized to strings.")


# ------------------------------------------------------------
# 3. Numeric preprocessing
# ------------------------------------------------------------
# Missing numeric values -> median

numeric_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        )
    ]
)


# ------------------------------------------------------------
# 4. Categorical preprocessing
# ------------------------------------------------------------
# Missing categorical values -> "Unknown"
# Then One-Hot Encoding.

categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="constant",
                fill_value="Unknown"
            )
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)


# ------------------------------------------------------------
# 5. Combine preprocessing
# ------------------------------------------------------------

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_pipeline,
            numeric_features
        ),
        (
            "cat",
            categorical_pipeline,
            categorical_features
        )
    ],
    remainder="drop"
)


# ------------------------------------------------------------
# 6. FIT ONLY ON TRAINING DATA
# ------------------------------------------------------------

X_train_processed = preprocessor.fit_transform(
    X_train_clean
)


# ------------------------------------------------------------
# 7. TRANSFORM TEST DATA
# ------------------------------------------------------------

X_test_processed = preprocessor.transform(
    X_test_clean
)


# ------------------------------------------------------------
# 8. Get feature names
# ------------------------------------------------------------

feature_names = preprocessor.get_feature_names_out()


# ------------------------------------------------------------
# 9. Convert processed data to DataFrame
# ------------------------------------------------------------

X_train_processed = pd.DataFrame(
    X_train_processed,
    columns=feature_names,
    index=X_train_clean.index
)

X_test_processed = pd.DataFrame(
    X_test_processed,
    columns=feature_names,
    index=X_test_clean.index
)


# ------------------------------------------------------------
# 10. VALIDATION
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("PREPROCESSING RESULTS")
print("=" * 60)

print("\nOriginal features:", X_train_clean.shape[1])

print(
    "Processed features:",
    X_train_processed.shape[1]
)

print(
    "\nX_train_processed shape:",
    X_train_processed.shape
)

print(
    "X_test_processed shape:",
    X_test_processed.shape
)

print(
    "\nMissing values in processed train:",
    X_train_processed.isna().sum().sum()
)

print(
    "Missing values in processed test:",
    X_test_processed.isna().sum().sum()
)


print("\n" + "=" * 60)
print("STEP 41 SUCCESS")
print("=" * 60)
```

```python
# ============================================================
# STEP 43: RANDOM FOREST PREDICTIONS
# ============================================================

print("=" * 60)
print("STEP 43: RANDOM FOREST PREDICTIONS")
print("=" * 60)


# ------------------------------------------------------------
# 1. Predict class
# ------------------------------------------------------------

y_pred = rf_model.predict(
    X_test_processed
)


# ------------------------------------------------------------
# 2. Predict probability of attrition
# ------------------------------------------------------------

y_pred_proba = rf_model.predict_proba(
    X_test_processed
)[:, 1]


# ------------------------------------------------------------
# 3. Basic validation
# ------------------------------------------------------------

print("\nPrediction completed successfully.")

print("\nPrediction shape:")
print("y_pred:", y_pred.shape)

print("\nProbability shape:")
print("y_pred_proba:", y_pred_proba.shape)


# ------------------------------------------------------------
# 4. Actual vs predicted distribution
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("ACTUAL TARGET DISTRIBUTION")
print("=" * 60)

print(
    y_test.value_counts().sort_index()
)


print("\n" + "=" * 60)
print("PREDICTED TARGET DISTRIBUTION")
print("=" * 60)

print(
    pd.Series(y_pred).value_counts().sort_index()
)


# ------------------------------------------------------------
# 5. Probability statistics
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("ATTRITION PROBABILITY SUMMARY")
print("=" * 60)

print(
    pd.Series(y_pred_proba).describe()
)


# ------------------------------------------------------------
# 6. Check probability range
# ------------------------------------------------------------

print("\nMinimum probability:", y_pred_proba.min())
print("Maximum probability:", y_pred_proba.max())


# ------------------------------------------------------------
# 7. Create prediction result DataFrame
# ------------------------------------------------------------

prediction_results = pd.DataFrame({
    "Actual_Attrition": y_test.values,
    "Predicted_Attrition": y_pred,
    "Attrition_Probability": y_pred_proba
})


print("\n" + "=" * 60)
print("SAMPLE PREDICTIONS")
print("=" * 60)

print(
    prediction_results.head(10)
)


print("\n" + "=" * 60)
print("STEP 43 SUCCESS")
print("=" * 60)
```

```python
# ============================================================
# STEP 45: ATTRITION PROBABILITY THRESHOLD ANALYSIS
# ============================================================

from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score
)

print("=" * 60)
print("STEP 45: ATTRITION PROBABILITY THRESHOLD ANALYSIS")
print("=" * 60)

thresholds = [
    0.50,
    0.40,
    0.30,
    0.25,
    0.20,
    0.15,
    0.10,
    0.05
]

threshold_results = []

for threshold in thresholds:

    y_pred_threshold = (
        y_pred_proba >= threshold
    ).astype(int)

    precision = precision_score(
        y_test,
        y_pred_threshold,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        y_pred_threshold,
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        y_pred_threshold,
        zero_division=0
    )

    predicted_attrition = int(
        y_pred_threshold.sum()
    )

    threshold_results.append({
        "Threshold": threshold,
        "Predicted_Attrition": predicted_attrition,
        "Precision": precision,
        "Recall": recall,
        "F1_Score": f1
    })


threshold_results_df = pd.DataFrame(
    threshold_results
)


print("\n" + "=" * 60)
print("THRESHOLD COMPARISON")
print("=" * 60)

print(
    threshold_results_df.to_string(
        index=False,
        formatters={
            "Precision": "{:.4f}".format,
            "Recall": "{:.4f}".format,
            "F1_Score": "{:.4f}".format
        }
    )
)


# ------------------------------------------------------------
# Best F1 threshold
# ------------------------------------------------------------

best_f1_row = threshold_results_df.loc[
    threshold_results_df["F1_Score"].idxmax()
]

print("\n" + "=" * 60)
print("BEST F1 THRESHOLD")
print("=" * 60)

print(
    f"Threshold           : "
    f"{best_f1_row['Threshold']:.2f}"
)

print(
    f"Predicted Attrition : "
    f"{int(best_f1_row['Predicted_Attrition'])}"
)

print(
    f"Precision           : "
    f"{best_f1_row['Precision']:.4f}"
)

print(
    f"Recall              : "
    f"{best_f1_row['Recall']:.4f}"
)

print(
    f"F1 Score            : "
    f"{best_f1_row['F1_Score']:.4f}"
)


print("\n" + "=" * 60)
print("STEP 45 SUCCESS")
print("=" * 60)
```

```python
# ============================================================
# STEP 47: AGGREGATED ORIGINAL FEATURE IMPORTANCE
# ============================================================

print("=" * 60)
print("STEP 47: AGGREGATED ORIGINAL FEATURE IMPORTANCE")
print("=" * 60)


# ------------------------------------------------------------
# 1. Copy feature importance
# ------------------------------------------------------------

agg_importance = feature_importance_df.copy()


# ------------------------------------------------------------
# 2. Extract original feature name
# ------------------------------------------------------------

def get_original_feature(feature_name):

    # Remove preprocessing prefix
    if "__" in feature_name:
        feature_name = feature_name.split("__", 1)[1]

    # For one-hot encoded categorical features,
    # remove the category after the first "_"
    #
    # Example:
    # Branch_NOIDA -> Branch
    # City Name_Noida -> City Name
    #
    # Numeric features remain unchanged.

    if feature_name.startswith("cat_"):
        feature_name = feature_name.replace("cat_", "", 1)

    return feature_name


# ------------------------------------------------------------
# IMPORTANT:
# Better method: use transformer output names
# and map based on the preprocessing transformers.
# ------------------------------------------------------------

numeric_features = list(
    preprocessor
    .named_transformers_["num"]
    .feature_names_in_
)

categorical_features = list(
    preprocessor
    .named_transformers_["cat"]
    .feature_names_in_
)


# ------------------------------------------------------------
# 3. Build mapping from processed feature -> original feature
# ------------------------------------------------------------

original_feature_mapping = {}

for feature in numeric_features:

    processed_name = f"num__{feature}"

    original_feature_mapping[processed_name] = feature


# Get encoded categorical feature names
ohe = (
    preprocessor
    .named_transformers_["cat"]
    .named_steps["onehot"]
)

encoded_names = ohe.get_feature_names_out(
    categorical_features
)

for encoded_name, original_feature in zip(
    encoded_names,
    [
        feature
        for feature in categorical_features
        for _ in range(
            len(
                ohe.categories_[
                    categorical_features.index(feature)
                ]
            )
        )
    ]
):

    processed_name = f"cat__{encoded_name}"

    original_feature_mapping[
        processed_name
    ] = original_feature


# ------------------------------------------------------------
# 4. Map every processed feature to original column
# ------------------------------------------------------------

agg_importance["Original_Feature"] = (
    agg_importance["Feature"]
    .map(original_feature_mapping)
)


# ------------------------------------------------------------
# 5. Check for unmapped features
# ------------------------------------------------------------

unmapped = agg_importance[
    agg_importance["Original_Feature"].isna()
]

print(
    f"\nUnmapped processed features: "
    f"{len(unmapped)}"
)


# ------------------------------------------------------------
# 6. Aggregate importance
# ------------------------------------------------------------

original_importance = (
    agg_importance
    .dropna(subset=["Original_Feature"])
    .groupby("Original_Feature")["Importance"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)


# ------------------------------------------------------------
# 7. Add percentage
# ------------------------------------------------------------

original_importance["Importance_Percentage"] = (
    original_importance["Importance"] * 100
)


# ------------------------------------------------------------
# 8. Display top 30 original features
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("TOP 30 ORIGINAL FEATURES")
print("=" * 60)

print(
    original_importance
    .head(30)
    .to_string(
        index=False,
        formatters={
            "Importance": "{:.6f}".format,
            "Importance_Percentage": "{:.2f}%".format
        }
    )
)


# ------------------------------------------------------------
# 9. Validation
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("AGGREGATED IMPORTANCE VALIDATION")
print("=" * 60)

print(
    f"Total aggregated importance: "
    f"{original_importance['Importance'].sum():.6f}"
)


# ------------------------------------------------------------
# 10. Save results
# ------------------------------------------------------------

original_importance.to_csv(
    "random_forest_original_feature_importance.csv",
    index=False
)

print(
    "\nSaved as:"
    " random_forest_original_feature_importance.csv"
)


print("\n" + "=" * 60)
print("STEP 47 SUCCESS")
print("=" * 60)
```

```python
# ============================================================
# STEP 49: CATEGORICAL FEATURE ATTRITION-RATE ANALYSIS
# ============================================================

print("=" * 60)
print("STEP 49: CATEGORICAL FEATURE ATTRITION-RATE ANALYSIS")
print("=" * 60)


# ------------------------------------------------------------
# 1. Top features from Step 47
# ------------------------------------------------------------

top_features = (
    original_importance
    .head(20)["Original_Feature"]
    .tolist()
)


# ------------------------------------------------------------
# 2. Identify categorical features
# ------------------------------------------------------------

categorical_features = [
    feature
    for feature in top_features
    if feature in model_data.columns
    and not pd.api.types.is_numeric_dtype(model_data[feature])
]


print("\nCategorical features selected:")
for i, feature in enumerate(categorical_features, 1):
    print(f"{i:02d}. {feature}")


# ------------------------------------------------------------
# 3. Minimum group size
# ------------------------------------------------------------

MIN_GROUP_SIZE = 20

print(
    f"\nMinimum employees required per category: "
    f"{MIN_GROUP_SIZE}"
)


# ------------------------------------------------------------
# 4. Calculate attrition rate by category
# ------------------------------------------------------------

all_category_results = []


for feature in categorical_features:

    temp = model_data[[feature, "Attrition"]].copy()

    # Handle missing values
    temp[feature] = temp[feature].fillna("Unknown")
    temp[feature] = temp[feature].astype(str)

    grouped = (
        temp
        .groupby(feature)["Attrition"]
        .agg(
            Employee_Count="count",
            Attrition_Count="sum",
            Attrition_Rate="mean"
        )
        .reset_index()
    )

    # Minimum group size
    grouped = grouped[
        grouped["Employee_Count"] >= MIN_GROUP_SIZE
    ].copy()

    if grouped.empty:
        continue

    grouped["Attrition_Rate"] = (
        grouped["Attrition_Rate"] * 100
    )

    grouped["Feature"] = feature

    grouped = grouped.rename(
        columns={feature: "Category"}
    )

    all_category_results.append(
        grouped[
            [
                "Feature",
                "Category",
                "Employee_Count",
                "Attrition_Count",
                "Attrition_Rate"
            ]
        ]
    )


# ------------------------------------------------------------
# 5. Combine results
# ------------------------------------------------------------

if all_category_results:

    categorical_attrition_analysis = pd.concat(
        all_category_results,
        ignore_index=True
    )

else:

    categorical_attrition_analysis = pd.DataFrame(
        columns=[
            "Feature",
            "Category",
            "Employee_Count",
            "Attrition_Count",
            "Attrition_Rate"
        ]
    )


# ------------------------------------------------------------
# 6. Sort by attrition rate
# ------------------------------------------------------------

categorical_attrition_analysis = (
    categorical_attrition_analysis
    .sort_values(
        by="Attrition_Rate",
        ascending=False
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 7. Display top categories
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("TOP CATEGORIES BY ATTRITION RATE")
print("=" * 60)

if not categorical_attrition_analysis.empty:

    display(
        categorical_attrition_analysis.head(30)
    )

else:

    print("No categorical results available.")


# ------------------------------------------------------------
# 8. Overall attrition rate
# ------------------------------------------------------------

overall_attrition_rate = (
    model_data["Attrition"].mean() * 100
)

print("\n" + "=" * 60)
print("OVERALL ATTRITION RATE")
print("=" * 60)

print(
    f"Overall attrition rate: "
    f"{overall_attrition_rate:.2f}%"
)


# ------------------------------------------------------------
# 9. Save results
# ------------------------------------------------------------

categorical_attrition_analysis.to_csv(
    "categorical_feature_attrition_analysis.csv",
    index=False
)


print("\nSaved as:")
print("categorical_feature_attrition_analysis.csv")


print("\n" + "=" * 60)
print("STEP 49 SUCCESS")
print("=" * 60)
```

```python

# ============================================================
# STEP 53: BUILD MISSING TRAINING FEATURES FOR PREDICTION
# ============================================================

print("=" * 60)
print("STEP 53: BUILDING MISSING TRAINING FEATURES")
print("=" * 60)

prediction_data = model_data.copy()

# ------------------------------------------------------------
# 1. ADD MISSING CATEGORICAL FEATURES
# ------------------------------------------------------------

missing_categorical = [
    "Point of Hire",
    "Point of Repatriation"
]

for col in missing_categorical:
    if col not in prediction_data.columns:
        prediction_data[col] = pd.NA
        print(f"Added missing categorical feature: {col}")


# ------------------------------------------------------------
# 2. DATE FEATURE ENGINEERING
# ------------------------------------------------------------

DATE_FEATURES = {
    "Actual Join Date": [
        "Actual Join Date_Days",
        "Actual Join Date_Missing"
    ],
    "Group Join Date": [
        "Group Join Date_Days",
        "Group Join Date_Missing"
    ],
    "Confirmation Date": [
        "Confirmation Date_Missing"
    ],
    "Effective Start Date": [
        "Effective Start Date_Days",
        "Effective Start Date_Missing"
    ],
    "Conf. Date (Period Master)": [
        "Conf. Date (Period Master)_Missing"
    ],
    "Birth Date": [
        "Birth Date_Days",
        "Birth Date_Missing"
    ],
    "Date Of Joined": [
        "Date Of Joined_Days",
        "Date Of Joined_Missing"
    ],
    "Created Date": [
        "Created Date_Days",
        "Created Date_Missing"
    ]
}


# ------------------------------------------------------------
# 3. FUNCTION TO CREATE DATE FEATURES
# ------------------------------------------------------------

def create_prediction_date_features(df, raw_column):

    # Convert to datetime
    dates = pd.to_datetime(
        df[raw_column],
        errors="coerce"
    )

    # --------------------------------------------------------
    # Treat clearly invalid employee dates as missing.
    # Effective Start Date contains year 0025 in the data.
    # --------------------------------------------------------
    invalid_dates = dates.notna() & (
        (dates.dt.year < 1900) |
        (dates.dt.year > 2100)
    )

    dates.loc[invalid_dates] = pd.NaT

    # Missing flag
    missing_flag = dates.isna().astype(int)

    # Days since Unix epoch
    days = (
        dates - pd.Timestamp("1970-01-01")
    ).dt.days

    return days, missing_flag


# ------------------------------------------------------------
# 4. CREATE ENGINEERED DATE FEATURES
# ------------------------------------------------------------

for raw_column, engineered_columns in DATE_FEATURES.items():

    print(f"\nProcessing: {raw_column}")

    days, missing_flag = create_prediction_date_features(
        prediction_data,
        raw_column
    )

    # Features ending in _Days
    for feature in engineered_columns:

        if feature.endswith("_Days"):
            prediction_data[feature] = days

        elif feature.endswith("_Missing"):
            prediction_data[feature] = missing_flag


# ------------------------------------------------------------
# 5. CHECK REQUIRED TRAINING FEATURES
# ------------------------------------------------------------

missing_after_engineering = [
    col
    for col in training_columns
    if col not in prediction_data.columns
]

extra_prediction_columns = [
    col
    for col in prediction_data.columns
    if col not in training_columns
]

print("\n" + "=" * 60)
print("FEATURE ALIGNMENT CHECK")
print("=" * 60)

print("Training feature count:", len(training_columns))
print("Prediction feature count:", len(prediction_data.columns))

print("\nMissing training features:")
print(len(missing_after_engineering))

if missing_after_engineering:
    for i, col in enumerate(missing_after_engineering, 1):
        print(f"{i:02d}. {col}")

print("\nExtra prediction columns:")
print(len(extra_prediction_columns))

if extra_prediction_columns:
    print(extra_prediction_columns)


# ------------------------------------------------------------
# 6. STOP IF ANY TRAINING FEATURES ARE STILL MISSING
# ------------------------------------------------------------

if missing_after_engineering:
    raise ValueError(
        "Prediction data is still missing training features. "
        "Do not continue."
    )


# ------------------------------------------------------------
# 7. ALIGN EXACTLY TO TRAINING ORDER
# ------------------------------------------------------------

prediction_data = prediction_data[
    training_columns
].copy()


# ------------------------------------------------------------
# 8. FINAL VALIDATION
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("FINAL PREDICTION FEATURE DATA")
print("=" * 60)

print("Shape:", prediction_data.shape)

print(
    "Expected shape:",
    (len(model_data), len(training_columns))
)

print("\nFirst 10 columns:")
print(prediction_data.columns[:10].tolist())

print("\nLast 10 columns:")
print(prediction_data.columns[-10:].tolist())

print("\nMissing values:")
print(prediction_data.isna().sum().sum())

print("\n" + "=" * 60)
print("STEP 53 SUCCESS")
print("=" * 60)
```

```python

# ============================================================
# STEP 54: FIND FITTED PREPROCESSOR
# ============================================================

print("=" * 60)
print("STEP 54: FINDING FITTED PREPROCESSOR")
print("=" * 60)

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

preprocessor_candidates = []

# IMPORTANT:
# Convert globals() to a list first so the dictionary
# cannot change while we iterate through it.

for name, obj in list(globals().items()):

    try:

        if isinstance(obj, ColumnTransformer):
            preprocessor_candidates.append(
                (name, type(obj).__name__)
            )

        elif isinstance(obj, Pipeline):
            preprocessor_candidates.append(
                (name, type(obj).__name__)
            )

    except Exception:
        pass


print("\nPreprocessing candidates found:")
print("-" * 60)

if not preprocessor_candidates:

    print("NO PREPROCESSOR FOUND")

else:

    for name, obj_type in preprocessor_candidates:

        print(f"Variable: {name}")
        print(f"Type:     {obj_type}")
        print("-" * 40)


print("\n" + "=" * 60)
print("STEP 54 COMPLETE")
print("=" * 60)
```

```python

# ============================================================
# STEP 55: TRANSFORM CURRENT EMPLOYEE DATA
# ============================================================

print("=" * 60)
print("STEP 55: TRANSFORMING PREDICTION DATA")
print("=" * 60)

# ------------------------------------------------------------
# 1. Make a fresh copy
# ------------------------------------------------------------

prediction_data_for_model = prediction_data.copy()

print("\nInput prediction data:")
print("Shape:", prediction_data_for_model.shape)


# ------------------------------------------------------------
# 2. Replace pandas pd.NA with numpy NaN
# ------------------------------------------------------------
# The fitted sklearn preprocessor can handle np.nan,
# but pd.NA can cause:
# "boolean value of NA is ambiguous"

prediction_data_for_model = prediction_data_for_model.replace(
    {pd.NA: np.nan}
)


# ------------------------------------------------------------
# 3. Make sure categorical columns use object/string-compatible
#    missing values rather than pd.NA
# ------------------------------------------------------------

if "categorical_features" in globals():

    for col in categorical_features:

        if col in prediction_data_for_model.columns:

            prediction_data_for_model[col] = (
                prediction_data_for_model[col]
                .astype(object)
                .where(
                    prediction_data_for_model[col].notna(),
                    np.nan
                )
            )

    print(
        "\nCategorical columns standardized:",
        len(categorical_features)
    )

else:

    print(
        "\nWarning: categorical_features variable not found."
    )


# ------------------------------------------------------------
# 4. Verify exact training feature order
# ------------------------------------------------------------

if list(prediction_data_for_model.columns) != list(training_columns):

    raise ValueError(
        "Prediction columns do not match training feature order."
    )

print("Feature order: VERIFIED")


# ------------------------------------------------------------
# 5. Check for pandas NA specifically
# ------------------------------------------------------------

pd_na_count = 0

for col in prediction_data_for_model.columns:

    try:
        pd_na_count += prediction_data_for_model[col].map(
            lambda x: x is pd.NA
        ).sum()

    except Exception:
        pass


print("Remaining pd.NA values:", pd_na_count)


if pd_na_count != 0:

    raise ValueError(
        "pd.NA values still exist in prediction data."
    )


# ------------------------------------------------------------
# 6. Apply the SAME fitted preprocessing pipeline
# ------------------------------------------------------------

print("\nApplying fitted preprocessor...")

X_prediction_processed = preprocessor.transform(
    prediction_data_for_model
)


# ------------------------------------------------------------
# 7. Check transformed shape
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("TRANSFORMATION RESULTS")
print("=" * 60)

print(
    "Processed prediction shape:",
    X_prediction_processed.shape
)

print(
    "Expected:",
    (len(prediction_data_for_model), 4354)
)


# ------------------------------------------------------------
# 8. Validate processed feature count
# ------------------------------------------------------------

if X_prediction_processed.shape[1] != 4354:

    raise ValueError(
        f"Expected 4354 processed features, "
        f"but got {X_prediction_processed.shape[1]}."
    )


# ------------------------------------------------------------
# 9. Check processed missing values
# ------------------------------------------------------------

if hasattr(X_prediction_processed, "toarray"):

    processed_array = X_prediction_processed.toarray()

else:

    processed_array = np.asarray(
        X_prediction_processed
    )


processed_missing = np.isnan(
    processed_array
).sum()


print(
    "Missing values after preprocessing:",
    processed_missing
)


# ------------------------------------------------------------
# 10. Final validation
# ------------------------------------------------------------

if processed_missing != 0:

    raise ValueError(
        "Processed prediction data still contains missing values."
    )


print("\n" + "=" * 60)
print("STEP 55 SUCCESS")
print("=" * 60)
```

```python

# ============================================================
# STEP 56: FUTURE RESIGNATION PREDICTION
# ============================================================

import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier

print("\n" + "=" * 60)
print("STEP 56: FUTURE RESIGNATION PREDICTION")
print("=" * 60)

# ------------------------------------------------------------
# 1. PREDICTION INPUT
# ------------------------------------------------------------

print("\nPREDICTION INPUT")
print("-" * 60)

print("Processed prediction shape:", X_prediction_processed.shape)
print("Expected shape: (20468, 4354)")

if X_prediction_processed.shape != (20468, 4354):
    raise ValueError(
        f"Unexpected prediction shape: "
        f"{X_prediction_processed.shape}"
    )

print("Prediction data shape: VERIFIED")

# ------------------------------------------------------------
# 2. FIND TRAINED RANDOM FOREST
# ------------------------------------------------------------

print("\nMODEL CHECK")
print("-" * 60)

rf_candidates = []

for name, obj in list(globals().items()):
    try:
        if isinstance(obj, RandomForestClassifier):
            rf_candidates.append((name, obj))
    except Exception:
        pass

print("Random Forest models found:")

for name, obj in rf_candidates:
    print(
        f"  {name} -> "
        f"{type(obj).__name__}, "
        f"trees={obj.n_estimators}, "
        f"features={obj.n_features_in_}"
    )

if len(rf_candidates) == 0:
    raise RuntimeError(
        "No RandomForestClassifier found. "
        "The trained model is not available."
    )

# If exactly one RF exists, use it.
# If multiple exist, prefer the one expecting 4354 features.

matching_models = [
    (name, obj)
    for name, obj in rf_candidates
    if getattr(obj, "n_features_in_", None) == 4354
]

if len(matching_models) == 1:
    model_name, rf_model = matching_models[0]

elif len(matching_models) > 1:
    raise RuntimeError(
        "Multiple Random Forest models expect 4354 features. "
        "Please identify the correct trained model."
    )

else:
    raise RuntimeError(
        "No Random Forest model found that expects "
        "4354 processed features."
    )

print("\nSelected trained model:", model_name)
print("Number of trees:", rf_model.n_estimators)
print("Model features:", rf_model.n_features_in_)

# ------------------------------------------------------------
# 3. VERIFY FEATURE COUNT
# ------------------------------------------------------------

if X_prediction_processed.shape[1] != rf_model.n_features_in_:
    raise ValueError(
        f"Feature mismatch: prediction has "
        f"{X_prediction_processed.shape[1]} features, "
        f"model expects {rf_model.n_features_in_}."
    )

print("\nFeature count: VERIFIED")

# ------------------------------------------------------------
# 4. GENERATE FUTURE RESIGNATION PROBABILITY
# ------------------------------------------------------------

print("\nGENERATING FUTURE RESIGNATION PROBABILITIES")
print("-" * 60)

prediction_probabilities = rf_model.predict_proba(
    X_prediction_processed
)[:, 1]

print(
    "Prediction probabilities generated:",
    len(prediction_probabilities)
)

print(
    "Minimum probability:",
    round(prediction_probabilities.min(), 4)
)

print(
    "Maximum probability:",
    round(prediction_probabilities.max(), 4)
)

print(
    "Mean probability:",
    round(prediction_probabilities.mean(), 4)
)

# ------------------------------------------------------------
# 5. APPLY BUSINESS THRESHOLD
# ------------------------------------------------------------

prediction_threshold = 0.50

predicted_attrition = (
    prediction_probabilities >= prediction_threshold
).astype(int)

print("\nPREDICTION THRESHOLD")
print("-" * 60)

print("Threshold:", prediction_threshold)

print(
    "Predicted future resignation cases:",
    int(predicted_attrition.sum())
)

print(
    "Predicted non-resignation cases:",
    int((predicted_attrition == 0).sum())
)

# ------------------------------------------------------------
# 6. CREATE RISK CATEGORY
# ------------------------------------------------------------

def get_risk_category(probability):

    if probability >= 0.50:
        return "High Risk"

    elif probability >= 0.20:
        return "Medium Risk"

    else:
        return "Low Risk"


risk_categories = [
    get_risk_category(p)
    for p in prediction_probabilities
]

# ------------------------------------------------------------
# 7. CREATE FINAL PREDICTION DATA
# ------------------------------------------------------------

prediction_results = model_data.copy()

# IMPORTANT:
# Historical Attrition is the target variable.
# It must NOT be presented as the future prediction.

if "Attrition" in prediction_results.columns:
    prediction_results = prediction_results.drop(
        columns=["Attrition"]
    )

prediction_results["Attrition_Probability"] = (
    prediction_probabilities
)

prediction_results["Attrition_Probability_%"] = (
    prediction_probabilities * 100
).round(2)

prediction_results["Predicted_Attrition"] = np.where(
    predicted_attrition == 1,
    "Yes",
    "No"
)

prediction_results["Risk_Category"] = risk_categories

# ------------------------------------------------------------
# 8. VERIFY RESULT
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("FUTURE RESIGNATION PREDICTION SUMMARY")
print("=" * 60)

print(
    "\nTotal employees evaluated:",
    len(prediction_results)
)

print(
    "Predicted future resignation:",
    int(predicted_attrition.sum())
)

print(
    "Predicted non-resignation:",
    int((predicted_attrition == 0).sum())
)

print("\nRISK DISTRIBUTION")
print("-" * 60)

print(
    prediction_results["Risk_Category"]
    .value_counts()
    .to_string()
)

# ------------------------------------------------------------
# 9. TOP 20 HIGH-RISK EMPLOYEES
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("TOP 20 EMPLOYEES BY FUTURE RESIGNATION RISK")
print("=" * 60)

top_risk_employees = (
    prediction_results
    .sort_values(
        "Attrition_Probability",
        ascending=False
    )
    .head(20)
)

# Find useful employee/business columns automatically
display_columns = []

possible_columns = [
    "Employee Code",
    "Employee ID",
    "Employee Name",
    "Name",
    "Title",
    "Designation",
    "Department",
    "Reporting Manager Name",
    "Site Name",
    "Cost Center",
    "Site Cost Center",
    "Client Vertical",
    "Monthly Gross Salary",
    "Annual_Gross_Salary",
    "Tenure_Years",
    "Age_Years",
    "Leave_Balance",
    "Attrition_Probability_%",
    "Predicted_Attrition",
    "Risk_Category"
]

for col in possible_columns:

    if col in top_risk_employees.columns:
        display_columns.append(col)

display(
    top_risk_employees[display_columns]
)

# ------------------------------------------------------------
# 10. SAVE FINAL PREDICTIONS
# ------------------------------------------------------------

prediction_results = prediction_results.sort_values(
    "Attrition_Probability",
    ascending=False
)

prediction_results.to_csv(
    "future_resignation_predictions.csv",
    index=False
)

print("\nSaved as:")
print("future_resignation_predictions.csv")

print("\n" + "=" * 60)
print("STEP 56 SUCCESS")
print("=" * 60)
```

```python

# ============================================================
# STEP 57: FINAL BUSINESS OUTPUT
# ============================================================

import pandas as pd
import numpy as np
import os

print("\n" + "=" * 70)
print("STEP 57: FINAL FUTURE RESIGNATION BUSINESS OUTPUT")
print("=" * 70)

# ------------------------------------------------------------
# 1. LOAD THE PREDICTION FILE CREATED IN STEP 56
# ------------------------------------------------------------

prediction_file = "future_resignation_predictions.csv"

if not os.path.exists(prediction_file):
    raise FileNotFoundError(
        f"{prediction_file} was not found. "
        "Please run Step 56 first."
    )

final_predictions = pd.read_csv(prediction_file)

print("\nPrediction file loaded:")
print("Shape:", final_predictions.shape)

# ------------------------------------------------------------
# 2. DISPLAY AVAILABLE COLUMNS
# ------------------------------------------------------------

print("\nAvailable prediction columns:")
print(final_predictions.columns.tolist())

# ------------------------------------------------------------
# 3. IDENTIFY PROBABILITY COLUMN
# ------------------------------------------------------------

probability_candidates = [
    "Future_Resignation_Probability",
    "Resignation_Probability",
    "Future_Resignation_Prob",
    "Prediction_Probability",
    "Probability",
    "Attrition_Probability"
]

probability_column = None

for col in probability_candidates:
    if col in final_predictions.columns:
        probability_column = col
        break

if probability_column is None:
    raise ValueError(
        "Could not identify the resignation probability column. "
        "Check the column list printed above."
    )

print("\nProbability column:", probability_column)

# ------------------------------------------------------------
# 4. IDENTIFY / CREATE RISK CATEGORY
# ------------------------------------------------------------

risk_candidates = [
    "Risk_Category",
    "Future_Resignation_Risk",
    "Risk",
    "Attrition_Risk"
]

risk_column = None

for col in risk_candidates:
    if col in final_predictions.columns:
        risk_column = col
        break

if risk_column is None:

    # Create risk category from probability
    def assign_risk(probability):
        if probability >= 0.50:
            return "High Risk"
        elif probability >= 0.20:
            return "Medium Risk"
        else:
            return "Low Risk"

    final_predictions["Risk_Category"] = (
        final_predictions[probability_column]
        .apply(assign_risk)
    )

    risk_column = "Risk_Category"

# ------------------------------------------------------------
# 5. STANDARDIZE PROBABILITY
# ------------------------------------------------------------

final_predictions[probability_column] = pd.to_numeric(
    final_predictions[probability_column],
    errors="coerce"
)

# If probability is stored as percentage (e.g. 86.88)
# convert it to decimal (0.8688)
if final_predictions[probability_column].max() > 1:
    final_predictions[probability_column] = (
        final_predictions[probability_column] / 100
    )

final_predictions["Resignation_Probability_%"] = (
    final_predictions[probability_column] * 100
).round(2)

# ------------------------------------------------------------
# 6. CREATE RISK RANK
# ------------------------------------------------------------

final_predictions = final_predictions.sort_values(
    by=probability_column,
    ascending=False
).reset_index(drop=True)

final_predictions["Risk_Rank"] = (
    final_predictions.index + 1
)

# ------------------------------------------------------------
# 7. CREATE FINAL BUSINESS DATASET
# ------------------------------------------------------------

# Keep useful employee/business columns if they exist
preferred_columns = [
    "Employee ID",
    "Employee_ID",
    "Employee Code",
    "Employee_Code",
    "Title",
    "Designation",
    "Department",
    "Reporting Manager Name",
    "Site Name",
    "Cost Center",
    "Site Cost Center",
    "Client Vertical",
    "Monthly Gross Salary",
    "Annual_Gross_Salary",
    "Tenure_Years",
    "Calculated_Tenure_Years",
    "Leave_Balance",
    "Leave_Consumed",
    "Leave_Future_Approved",
    "OT1_Total_OT",
    "OT2_Total_OT",
    "OT3_Total_OT",
    "OT4_Total_OT",
    "Salary_Data_Available",
    "Leave_Data_Available",
    "Resignation_Probability_%",
    "Risk_Category",
    "Risk_Rank"
]

available_columns = [
    col for col in preferred_columns
    if col in final_predictions.columns
]

final_business_output = final_predictions[available_columns].copy()

# ------------------------------------------------------------
# 8. IF NO EMPLOYEE ID EXISTS, PRESERVE ORIGINAL ROW ID
# ------------------------------------------------------------

id_columns = [
    "Employee ID",
    "Employee_ID",
    "Employee Code",
    "Employee_Code"
]

if not any(col in final_business_output.columns for col in id_columns):

    # Use original prediction row number as reference
    final_business_output.insert(
        0,
        "Employee_Row_ID",
        np.arange(1, len(final_business_output) + 1)
    )

# ------------------------------------------------------------
# 9. FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL FUTURE RESIGNATION PREDICTION SUMMARY")
print("=" * 70)

print(
    "\nTotal employees evaluated:",
    len(final_business_output)
)

print(
    "Predicted High Risk:",
    (final_predictions[risk_column] == "High Risk").sum()
)

print(
    "Predicted Medium Risk:",
    (final_predictions[risk_column] == "Medium Risk").sum()
)

print(
    "Predicted Low Risk:",
    (final_predictions[risk_column] == "Low Risk").sum()
)

print(
    "\nAverage resignation probability:",
    round(
        final_predictions[probability_column].mean() * 100,
        2
    ),
    "%"
)

print(
    "Maximum resignation probability:",
    round(
        final_predictions[probability_column].max() * 100,
        2
    ),
    "%"
)

# ------------------------------------------------------------
# 10. RISK DISTRIBUTION
# ------------------------------------------------------------

print("\nRISK DISTRIBUTION")
print("-" * 50)

risk_distribution = (
    final_predictions[risk_column]
    .value_counts()
    .reindex(
        ["High Risk", "Medium Risk", "Low Risk"],
        fill_value=0
    )
)

print(risk_distribution)

# ------------------------------------------------------------
# 11. TOP 20 HIGH-RISK EMPLOYEES
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("TOP 20 EMPLOYEES BY FUTURE RESIGNATION RISK")
print("=" * 70)

display(
    final_business_output.head(20)
)

# ------------------------------------------------------------
# 12. SAVE FINAL BUSINESS OUTPUT
# ------------------------------------------------------------

output_file = "final_future_resignation_predictions.csv"

final_business_output.to_csv(
    output_file,
    index=False
)

print("\n" + "=" * 70)
print("FINAL OUTPUT FILE CREATED")
print("=" * 70)

print("\nFile:", output_file)
print("Rows:", len(final_business_output))
print("Columns:", len(final_business_output.columns))

print("\nSTEP 57 SUCCESS")
print("=" * 70)
```
